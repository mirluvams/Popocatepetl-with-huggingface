## Cargar modelos locales preentrenados y subirlos a Hugging Face! :)

```python
from transformers import AutoModelForImageClassification
```

```python
model_names = [
    "microsoft/swinv2-base-patch4-window16-256",
    "microsoft/beit-base-patch16-224",
    "google/vit-base-patch16-224",
    "facebook/convnext-base",
    "facebook/dinov2-base"
]
```

```python
hub_user = "mirluvams"
project_name = "popocatepetl"
```

```python
for model_name in model_names:
    print(f"Subiendo {model_name}...")

    model = AutoModelForImageClassification.from_pretrained(model_name)
    
    short_name = model_name.split("/")[-1]
    repo_name = f"{hub_user}/{short_name}-{project_name}"

    model.push_to_hub(repo_name, private=True)

    print(f"Subido a https://huggingface.co/{repo_name}\n")
```
